### Input data paths

```python
datapath = "data/d921c710-4e4d-4316-96c5-666600485eb7"
filename = "UK1_LS_RF_Mean_1980-2018_COG.tif"
datapath1 = os.path.join(datapath, filename)

filename = "UK1_LS_RF_Mean_1980-2018_COG.tif"
datapath2 = os.path.join(datapath, filename)
```

 ### Load raster data

```python
with rasterio.open(datapath1) as src:
    # Read all bands as a NumPy array: (bands, height, width)
    img = src.read()
    # Geographic bounds of raster: (left, bottom, right, top)
    bounds = src.bounds
```

### Format image for display

```python
# Convert from (bands, H, W) → (H, W, bands)
img = np.moveaxis(img, 0, -1)

# If single-band raster, squeeze to 2D for folium compatibility
if img.shape[2] == 1:
    img = img[:, :, 0]
```

### Initialize map centered on raster extent

```python
center_lat = (bounds.top + bounds.bottom) / 2
center_lon = (bounds.left + bounds.right) / 2

m = folium.Map(location=[center_lat, center_lon])
```

### Overlay raster on map

```python
folium.raster_layers.ImageOverlay(
    image=img,
    bounds=[
        [bounds.bottom, bounds.left],  # south-west corner
        [bounds.top, bounds.right]     # north-east corner
    ],
    opacity=0.6,
    name="Raster Layer"
).add_to(m)
```
